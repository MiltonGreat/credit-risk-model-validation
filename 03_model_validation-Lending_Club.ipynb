## 03_model_validation

## Reading this notebook

Notebook 02 built the models. This notebook validates them — independently, and
with the expectation that the answer might be "the model is not fit for purpose."

**What "independent validation" means here:**

Independent validation is not a re-run of the development notebook. It is a
separate assessment that loads the trained models, evaluates them against
standards the developer did not set, and reports findings the developer might
not like. The output of this notebook is not a model — it is a *judgment* about
a model.

Three things distinguish this notebook from Notebook 02:

1. **It loads the models rather than retraining them.** Validation must assess
   what was actually built, not a fresh equivalent.
2. **It compares against expectations set in advance.** The realistic AUC of
   0.621 was established in Notebook 02; this notebook checks whether the models
   reproduce it.
3. **It is willing to conclude that the model should not be deployed.** And in
   fact it does — the risk rating is HIGH and the recommendation is REJECT.

**The phases:**

- Phases 0–3: setup and independent loading of models and validation data
- Phase 4: leakage impact (reproduced and confirmed)
- Phase 5: performance validation (does 0.621 reproduce?)
- Phase 6: stability (PSI — with a documented fallback)
- **Phase 7: SHAP explainability and monotonicity** (the conceptual soundness check)
- Phase 8: calibration assessment
- Phase 9: calibration plot and KS statistic
- **Phase 10: fairness testing** (the finding that drives the risk rating)
- Phase 11: validation findings register (12 findings)
- Phase 12: risk score and rating rationale
- Phases 13–14: validation report and limitations

**Where the notebook is strongest, and where it needs care:**

The strongest part is the findings register and the risk rating rationale — this

### PHASE 5: PERFORMANCE VALIDATION

### Why this phase exists

Four models are evaluated on the validation set, and the results are compared

**What the output shows:**

- The realistic AUC expectation is reproduced exactly (0.621, XGBoost).
- Performance Rating: MODERATE.

**How to read "MODERATE" carefully:** The rating is set by a simple band —
AUC > 0.7 is GOOD, > 0.6 is MODERATE, below is POOR. An AUC of 0.621 is at the
low end of MODERATE. It is honest to call it moderate; it would be misleading to
call it good.

**The tension to carry forward:** This phase rates the model's *performance* as
MODERATE. Phase 12 rates the *model* as HIGH risk and recommends REJECT. Both are
correct — performance is one component of the overall assessment, and a model can
have acceptable performance and still be unfit for deployment. The notebook does
not currently state this connection explicitly; a reader will have to infer it.
Adding one sentence — "performance alone does not determine deployability; the
overall risk rating incorporates fairness, calibration, and stability as well" —
would close the gap.

```python
print("\n" + "=" * 80)
print("PHASE 5: PERFORMANCE VALIDATION (V2.0)")
print("=" * 80)

def validate_performance(y_true, y_proba, name):
    fpr, tpr, thresholds = roc_curve(y_true, y_proba)
    youden_j = tpr - fpr
    optimal_idx = np.argmax(youden_j)
    threshold = thresholds[optimal_idx] if len(thresholds) > 0 else 0.5
    
    y_pred = (y_proba >= threshold).astype(int)
    
    return {
        'Model': name,
        'AUC-ROC': roc_auc_score(y_true, y_proba),
        'PR-AUC': average_precision_score(y_true, y_proba),
        'Brier': brier_score_loss(y_true, y_proba),
        'Log_Loss': log_loss(y_true, y_proba),
        'Accuracy': accuracy_score(y_true, y_pred),
        'Precision': precision_score(y_true, y_pred, zero_division=0),
        'Recall': recall_score(y_true, y_pred, zero_division=0),
        'F1': f1_score(y_true, y_pred, zero_division=0),
        'Threshold': threshold
    }

validation_results = []
for name in models.keys():
    metrics = validate_performance(y_val, predictions[name], name)
    validation_results.append(metrics)

val_df = pd.DataFrame(validation_results)
print("\nVALIDATION RESULTS (V2.0):")
print(val_df.round(4).to_string(index=False))

best_model = val_df.loc[val_df['AUC-ROC'].idxmax(), 'Model']
best_auc = val_df['AUC-ROC'].max()

print(f"\nBest Model: {best_model} (AUC-ROC: {best_auc:.4f})")

# V2.0: Performance validation against expectations
print("\n" + "=" * 80)
print("PERFORMANCE VALIDATION (V2.0)")
print("=" * 80)

print(f"Expected Realistic AUC: {realistic_auc:.3f}")
print(f"Validated AUC:          {best_auc:.3f}")
print(f"Difference:            {best_auc - realistic_auc:.3f}")
if abs(best_auc - realistic_auc) < 0.05:
    print("[OK] Performance replicated successfully")
else:
    print("[WARNING] Performance deviation detected")

if best_auc > 0.7:
    perf_rating = "GOOD"
elif best_auc > 0.6:
    perf_rating = "MODERATE"
else:
    perf_rating = "POOR"

print(f"Performance Rating: {perf_rating}")

print("""
Note on interpretation:
----------------------
A MODERATE performance rating does not by itself determine deployability.
The overall model risk rating (Phase 12) incorporates five components:
performance, stability, calibration, fairness, and leakage.

A model can perform moderately and still be unfit for deployment if, for
example, it treats groups unequally or produces uncalibrated probabilities.
That is the case here. See Phase 12 for the overall rating and its rationale.
""")
```

### PHASE 6: STABILITY ANALYSIS (PSI)

### Why this phase exists

PSI measures whether the distribution of the model's predictions has shifted
between two populations — in principle, two time periods. A large PSI means the
model is seeing a different world than the one it was trained on, which is a
leading indicator of performance decay.

**What the phase claims to do:** compute PSI on time-based cohorts (early
vintages vs. later vintages).

**What the phase actually does in this run:** falls back to an index-based split,
because no date column is available. The output states this clearly:

    [WARNING] Using index-based cohorts (data may not be time-sorted)

**Why that matters:** an index-based PSI is essentially a random split. A random
split of a stable population will produce a PSI near zero — which is exactly what
the output shows (PSI = 0.0098, "Stable"). The PSI is stable *because the split
is arbitrary*, not because the population is stable over time. The notebook is
honest about the fallback, but a reader who only sees "Stable (PSI < 0.1)" will
draw a stronger conclusion than the data supports.

**The governance lesson:** a fallback can be worse than a null result, because it
produces a number that looks like evidence. If the fallback cannot answer the
question the phase is asking, the phase should report "not assessed" rather than
report a misleading figure. This is a candidate for a future revision.

```python
print("\n" + "=" * 80)
print("PHASE 6: POPULATION STABILITY INDEX (PSI) — TIME COHORTS")
print("=" * 80)

"""
DECISION POINT 5: PSI COHORT DEFINITION
========================================
Decision: Split by time cohorts (vintage years)

Rationale:
1. Time-based split is more meaningful for stability analysis
2. Detects drift between origination cohorts
3. Regulatory expectation (OSFI E-23 Section 3.2)

Cohorts:
- Cohort 1: Loans from earlier vintage
- Cohort 2: Loans from later vintage

Thresholds:
- PSI < 0.10: STABLE (no significant shift)
- PSI 0.10-0.25: MODERATE SHIFT (monitor)
- PSI > 0.25: MAJOR SHIFT (investigate)

Regulatory Reference:
- OSFI E-23 Section 3.2: Performance Monitoring (population stability)
- SR 11-7: Ongoing monitoring
"""

def calculate_psi(expected, actual, bins=10):
    """Calculate Population Stability Index (PSI)."""
    bin_edges = np.linspace(0, 1, bins+1)
    expected_hist, _ = np.histogram(expected, bins=bin_edges)
    actual_hist, _ = np.histogram(actual, bins=bin_edges)
    
    expected_hist = expected_hist + 1e-10
    actual_hist = actual_hist + 1e-10
    
    expected_prop = expected_hist / expected_hist.sum()
    actual_prop = actual_hist / actual_hist.sum()
    
    psi = np.sum((expected_prop - actual_prop) * np.log(expected_prop / actual_prop))
    return psi

psi = None
psi_status = "N/A"

if best_model in predictions:
    y_proba = predictions[best_model]
    
    # Try time-based cohorts if vintage information available
    # Note: This requires that the validation data has vintage information
    # If not available, use index-based split as fallback
    
    # Check if we have vintage year in the data
    # For Lending Club, we might have issue_month or similar
    # For now, use index-based split with clear documentation
    n = len(y_proba)
    half = n // 2
    
    # Check if data is sorted by time (if date_col_used was provided)
    if split_type == 'time-based':
        # Assume data is already in time order
        proba_early = y_proba[:half]
        proba_late = y_proba[half:]
        print("\nUsing time-based cohorts (data sorted by origination date)")
    else:
        # Fallback: split by index with warning
        proba_early = y_proba[:half]
        proba_late = y_proba[half:]
        print("\n[WARNING] Using index-based cohorts (data may not be time-sorted)")
    
    psi = calculate_psi(proba_early, proba_late)
    
    print(f"\nPSI for {best_model}: {psi:.4f}")
    
    if psi < 0.10:
        psi_status = "STABLE"
        print("[OK] Stable (PSI < 0.1)")
    elif psi < 0.25:
        psi_status = "MODERATE SHIFT"
        print("[WARNING] Moderate shift")
    else:
        psi_status = "MAJOR SHIFT"
        print("[ERROR] Major shift")
```

### PHASE 7: SHAP EXPLAINABILITY WITH DEPENDENCE PLOTS AND MONOTONICITY CHECK

### Why this phase exists

This is the conceptual soundness check. Two things are being asked:

1. **Which features drive the model's predictions?** — via SHAP importance.
2. **Do those features behave in the direction credit risk theory expects?** —
   via the monotonicity check. FICO should reduce default risk; DTI should
   increase it. A model that gets these backwards may be fitting noise.

**What the output shows:**

- Top features: `interest_rate`, `debt_to_income`, `loan_to_income`,
  `inquiries_last_12m`, `total_debit_limit`.
- **The SHAP summary plot failed:** `summary_legacy() got an unexpected keyword
  argument 'ax'` — a library version incompatibility.
- The dependence plots and monotonicity check are inside the same `try` block,
  so they did not run either.
- The leakage check emitted a **false positive:** `num_cc_carrying_balance` was
  flagged because it contains the substring "balance," but it is a credit-bureau
  feature available at origination, not one of the nine leakage features.

**The consequences, stated plainly:**

- **Finding 12 reports "0 features showed counterintuitive relationships."** But
  the check never ran. Reporting zero findings from a check that did not execute
  is a false negative, and it is the most serious correctness issue in this
  notebook.
- The leakage warning is misleading. The pattern list that was appropriate in
  Notebook 01 (where it triggered human review) is too loose here, where it
  produces an incorrect warning.

**What to take from this:** the SHAP importance values that did print are
plausible and directionally sensible — `interest_rate` and `debt_to_income` at
the top is what a credit model should look like. But the conceptual soundness
*check* — the monotonicity test — did not run. The notebook should not claim to
have assessed monotonicity until it actually has.

### PHASE 8: CALIBRATION ASSESSMENT

### Why this phase exists

A model can rank well and still produce probabilities that are systematically
wrong. This phase assesses calibration — whether a predicted probability of 0.10
corresponds to roughly a 10% observed default rate.

**What the output shows:**

- Logistic Regression: HL p-value 1.0000 — good calibration
- Random Forest: HL p-value 1.0000 — good calibration
- LightGBM: HL p-value 0.2972 — acceptable
- **XGBoost: HL p-value 0.0000 — poor calibration**

**Why this matters, and why it is easy to miss:** XGBoost is the best model by
AUC (0.621). It is also the worst-calibrated. The notebook's Phase 5 rating
("Best Model: XGBoost") is therefore only correct if "best" means "best ranker."
If "best" means "most usable for decision-making," XGBoost is the wrong choice
until its calibration is corrected.

**The governance significance:** calibration is not a secondary concern. It
matters whenever the model's probabilities are used quantitatively — pricing,
provisioning, capital allocation. A model that ranks well but is miscalibrated
can produce losses even when its ordering is correct. The notebook reports this
finding accurately; a reviewer should not lose sight of it when reading the
headline AUC.

```python
print("\n" + "=" * 80)
print("PHASE 8: CALIBRATION ASSESSMENT WITH PLOT AND KS STATISTIC")
print("=" * 80)

def assess_calibration_detailed(y_true, y_proba, name, n_bins=10):
    fraction_positive, mean_predicted = calibration_curve(
        y_true, y_proba, n_bins=n_bins, strategy='quantile'
    )
    
    bins = np.percentile(y_proba, np.linspace(0, 100, n_bins+1))
    bin_indices = np.digitize(y_proba, bins[1:-1])
    
    observed = []
    expected = []
    for i in range(n_bins):
        mask = bin_indices == i
        if mask.sum() > 0:
            observed.append(y_true[mask].sum())
            expected.append(y_proba[mask].sum())
    
    hl_stat = 0
    for obs, exp in zip(observed, expected):
        if exp > 0 and exp < len(observed):
            hl_stat += ((obs - exp) ** 2) / (exp * (1 - exp/len(observed)))
    
    df = n_bins - 2
    p_value = 1 - chi2.cdf(hl_stat, df) if df > 0 else 1.0
    
    cal_model = LogisticRegression()
    cal_model.fit(y_proba.reshape(-1, 1), y_true)
    cal_slope = cal_model.coef_[0][0]
    cal_intercept = cal_model.intercept_[0]
    
    return {
        'Model': name,
        'Calibration_Slope': cal_slope,
        'Calibration_Intercept': cal_intercept,
        'HL_Statistic': hl_stat,
        'HL_p_value': p_value,
        'Mean_Predicted': mean_predicted,
        'Fraction_Positive': fraction_positive
    }

calibration_results = []
for name in models.keys():
    cal_result = assess_calibration_detailed(y_val, predictions[name], name)
    calibration_results.append(cal_result)

# Create calibration metrics DataFrame
cal_metrics_df = pd.DataFrame([{
    'Model': r['Model'],
    'Slope': r['Calibration_Slope'],
    'Intercept': r['Calibration_Intercept'],
    'HL_pvalue': r['HL_p_value']
} for r in calibration_results])

print("\nCALIBRATION METRICS:")
print("Perfect calibration: Slope=1, Intercept=0, HL p-value > 0.05")
print("=" * 60)

for result in calibration_results:
    print(f"\n{result['Model']}:")
    print(f"  Calibration Slope: {result['Calibration_Slope']:.4f} (target: 1.0)")
    print(f"  Calibration Intercept: {result['Calibration_Intercept']:.4f} (target: 0.0)")
    print(f"  Hosmer-Lemeshow p-value: {result['HL_p_value']:.4f} (target: > 0.05)")
    
    if result['HL_p_value'] > 0.05:
        print("  [OK] Good calibration")
    else:
        print("  [WARNING] Poor calibration")
```

**Perfect Calibration:**

- Slope = 1.0 (predictions match reality)
- Intercept = 0.0 (no systematic bias)
- HL p-value > 0.05 (no significant miscalibration)

HL p-value	>0.05	0.000	Miscalibration confirmed

### PHASE 10: FAIRNESS TESTING (V2.0 - ENHANCED)

### Why this phase exists

This is the phase that determines the model's deployment status. It tests
whether the model approves or flags different groups at different rates, using
state as a geographic proxy for protected class.

**What the output shows:** severe disparate impact across the states tested.

- `state_AL`: DI = 0.5279
- `state_AR`: DI = 0.5554
- `state_AZ`: DI = 0.2435

All three are below the 0.6 threshold for "severe." The minimum DI is 0.2435.

**What DI below 0.6 means:** the least-approved group is approved at less than
60% the rate of the most-approved group. At 0.2435, the `state_AZ` group is
approved at roughly one-quarter the rate of the reference group. Under the four-
fifths rule (the standard in U.S. fair lending law), this would be treated as
evidence of disparate impact requiring justification or mitigation.

**The three caveats the notebook itself states, and that a reader should hold:**

1. **State is an imperfect proxy.** A geographic disparity is not the same as a
   racial or ethnic disparity. The DI figures are indicative, not definitive.
2. **Only three states were tested.** The loop limits to `protected_cols[:3]` for
   performance. The reported minimum is the minimum *of those three*, not of all
   states. The notebook should say so.
3. **The model has not been mitigated yet.** This is the finding that drives the
   remediation work in Notebook 04 and the cross-dataset test in Notebook 05.
   The point of the fairness test is to *find* the problem so it can be fixed.

**Why this finding is the most important one in the notebook:** it is what turns
the validation verdict from "moderate performance, some limitations" into
"deployment blocked." The model's performance is not disqualifying. Its fairness
profile is.

```python
print("\n" + "=" * 80)
print("PHASE 10: FAIRNESS TESTING WITH PROTECTED CLASS PROXY (V2.0)")
print("=" * 80)

"""
===============================================================================
FAIRNESS TESTING — PROTECTED CLASS PROXY DOCUMENTATION (V2.0)
===============================================================================

DECISION POINT 7: PROTECTED CLASS PROXY
========================================
Decision: Use state as a geographic proxy for protected class analysis

Rationale:
1. No explicit race/gender data available in Lending Club dataset
2. Geographic location (state) is a recognized proxy for demographic diversity
3. State-level analysis is a common approach in fair lending testing

Protected Class Proxied:
- Race/Ethnicity (geographic proxy)
- Geographic diversity

Limitation:
- State is an imperfect proxy for race/ethnicity
- Results should be interpreted as indicative, not definitive
- Additional testing recommended with explicit protected attributes

Regulatory Reference:
- ECOA: Fair lending requires testing for disparate impact
- CFPB: Geographic proxies are acceptable for fair lending testing
- OSFI E-23 Section 5.2: Fairness/Bias testing

Severity Justification:
- DI < 0.6 is SEVERE disparate impact
- Indicates potential fair lending violation
- Requires immediate mitigation===============================================================================
"""

print("\nProtected Class Proxy Documentation:")
print("  - Proxy Variable: State (geographic location)")
print("  - Protected Class Proxied: Race/Ethnicity, Geographic Diversity")
print("  - Threshold: 4/5ths Rule (80%)")
print("  - Severity Classification: DI < 0.6 = SEVERE, 0.6-0.8 = MODERATE, > 0.8 = NONE")
print()
print("Regulatory Reference: ECOA, CFPB, OSFI E-23 Section 5.2")

def comprehensive_fairness_analysis(X_data, y_true, y_pred, model):
    fairness_report = []
    
    protected_cols = []
    for col in X_data.columns:
        if any(term in col.lower() for term in ['state', 'gender', 'sex', 'race', 'ethnicity']):
            if X_data[col].nunique() <= 20:
                protected_cols.append(col)
    
    if not protected_cols:
        np.random.seed(RANDOM_STATE)
        X_data['Gender_Synthetic'] = np.random.choice([0, 1], size=len(X_data), p=[0.6, 0.4])
        protected_cols = ['Gender_Synthetic']
        print("  [NOTE] No protected attributes found. Using synthetic gender for demonstration.")
    
    print(f"\n  Found protected attributes: {protected_cols[:5]}... (showing first 5)")
    
    for attr in protected_cols[:3]:  # Limit to first 3 for performance
        groups = X_data[attr].unique()
        group_metrics = []
        
        for group in groups[:5]:
            mask = X_data[attr] == group
            if mask.sum() > 10:
                y_true_group = y_true[mask]
                y_pred_group = y_pred[mask]
                
                tp = ((y_true_group == 1) & (y_pred_group == 1)).sum()
                fp = ((y_true_group == 0) & (y_pred_group == 1)).sum()
                fn = ((y_true_group == 1) & (y_pred_group == 0)).sum()
                tn = ((y_true_group == 0) & (y_pred_group == 0)).sum()
                
                tpr = tp / (tp + fn) if (tp + fn) > 0 else 0
                fpr = fp / (fp + tn) if (fp + tn) > 0 else 0
                precision = tp / (tp + fp) if (tp + fp) > 0 else 0
                
                group_metrics.append({
                    'Attribute': attr,
                    'Group': str(group)[:20],
                    'N': mask.sum(),
                    'TPR': tpr,
                    'FPR': fpr,
                    'Precision': precision,
                    'Approval_Rate': y_pred_group.mean(),
                    'Default_Rate': y_true_group.mean()
                })
        
        if len(group_metrics) >= 2:
            g_df = pd.DataFrame(group_metrics)
            print(f"\n  Fairness Analysis by {attr}:")
            print(g_df.round(4).to_string(index=False))
            
            min_approval = g_df['Approval_Rate'].min()
            max_approval = g_df['Approval_Rate'].max()
            di_ratio = min_approval / max_approval if max_approval > 0 else 0
            
            print(f"\n  Disparate Impact Ratio: {di_ratio:.4f}")
            if di_ratio > 0.8:
                print("    [OK] No disparate impact (>= 0.8)")
            elif di_ratio > 0.6:
                print("    [WARNING] Monitor disparate impact (0.6-0.8)")
            else:
                print("    [ERROR] Severe disparate impact (< 0.6) - mitigation required")
            
            fairness_report.append({
                'Attribute': attr,
                'DI_Ratio': di_ratio,
                'Group_Count': len(g_df)
            })
    
    return pd.DataFrame(fairness_report) if fairness_report else pd.DataFrame()

print("\nFAIRNESS ASSESSMENT:")
print("=" * 50)

best_model_instance = models[best_model]
y_pred_best = best_model_instance.predict(X_val)

fairness_df = comprehensive_fairness_analysis(X_val, y_val, y_pred_best, best_model_instance)

# V2.0: Enhanced fairness documentation
print("\n" + "=" * 80)
print("FAIRNESS VALIDATION FINDINGS (V2.0)")
print("=" * 80)

if not fairness_df.empty:
    min_di = fairness_df['DI_Ratio'].min()
    print(f"\nMINIMUM DISPARATE IMPACT: {min_di:.4f}")
    print(f"  (Note: analysis limited to {len(fairness_df)} protected attributes "
          f"for performance. Minimum DI is the minimum over those {len(fairness_df)} "
          f"attributes, not all states.)")
    
    if min_di < 0.6:
        print("\n[CRITICAL] SEVERE DISPARATE IMPACT CONFIRMED")
        print("  - DI < 0.6 in multiple states")
        print("  - Model CANNOT be deployed without fairness mitigation")
        print("\nRECOMMENDED MITIGATION:")
        print("  1. Remove state features from model")
        print("  2. Apply reweighting (Kamiran & Calders)")
        print("  3. Apply fairness constraints during training")
        
        fairness_status = "CRITICAL - DEPLOYMENT BLOCKED"
    elif min_di < 0.8:
        fairness_status = "MONITOR - Further investigation required"
    else:
        fairness_status = "OK - No disparate impact detected"
    
    print(f"\nFairness Status: {fairness_status}")
    
    print("\nFAIRNESS SUMMARY:")
    print(fairness_df.round(4).to_string(index=False))
```

#### What It Means

This independently tests whether the model has disparate impact across protected classes. Since Lending Club doesn't have race/gender data, state is used as a geographic proxy for race/ethnicity.

#### Explain the Decision
- **State as proxy for race/ethnicity** Geographic location is a recognized proxy for demographic diversity. This is an acceptable approach under ECOA/CFPB guidance.
- **4/5ths rule (80% threshold)** The ECOA standard — if one group's approval rate is <80% of another group's rate, it's evidence of adverse impact.
- **DI: 0.2435 (SEVERE)** The minimum Disparate Impact ratio is 0.2435 — far below 0.8. This is SEVERE disparate impact.
- **Multiple states show severe DI** state_AL: 0.5279, state_AR: 0.5554, state_AZ: 0.2435 — all below 0.6.
- **DEPLOYMENT BLOCKED** This is a CRITICAL finding — the model cannot be deployed without fairness mitigation.

**I independently validate fairness. The model has severe disparate impact (DI 0.2435). This is a deployment blocker.**

### PHASE 12: RISK SCORE AND MODERATE-HIGH RATING RATIONALE

### Why this phase exists

This is where the validation verdict is produced. The model is scored across
five components — performance, stability, calibration, fairness, leakage — and
the total determines a risk rating and a deployment recommendation.

**What the output shows:** risk score 6/10, risk rating **HIGH**, recommendation
**REJECT**.

**The problem with this phase as currently written:** the phase title and the
rationale block both say "MODERATE-HIGH," but the rating scale printed
immediately above them defines 5–6 as HIGH. The rationale then argues "why
MODERATE-HIGH (6/10) rather than MODERATE (2/10) or HIGH (6/10)" — using
MODERATE-HIGH as a label for a score that is, by its own scale, HIGH. And
"why not HIGH (6/10)" is logically empty: the score *is* 6, which *is* HIGH.

**What is actually happening:** the underlying analysis is sound. The score is
6, the rating is HIGH, the recommendation is REJECT. But the labels were not
updated when the score changed, so the notebook's own title contradicts its own
output. A reader who notices this will reasonably wonder what else is out of sync.

**The fix is straightforward:** rename the phase and the rationale block to
"HIGH RATING RATIONALE," and rewrite the comparative argument as "why HIGH rather
than CRITICAL (8/10)" and "why HIGH rather than MODERATE (4/10)." The score does
not need to change; only the label does.

**Why the HIGH rating is defensible:** of the 6 points, 3 come from severe
fairness concerns and 2 from poor calibration. Performance contributes 1 and
stability 0. The rating is HIGH largely because of fairness and calibration, not
because of weak performance. That is the correct weighting for a deployment
decision — a model that performs moderately but treats groups unequally should
not be deployed, regardless of its AUC.

**The conclusion the notebook reaches — and it is the right one:** the model may
support human-led collections prioritization, but it cannot be deployed for
fairness-sensitive decisions, IFRS 9 provisioning, Canadian portfolios, or
autonomous decisions. That is a specific, actionable verdict, and it is exactly
what a validation notebook should produce.

```python
print("\n" + "=" * 80)
print("PHASE 12: RISK SCORE AND HIGH RATING RATIONALE")
print("=" * 80)

"""
DECISION POINT 9: RISK SCORING METHODOLOGY
===========================================
Decision: Risk score calculated on a 0-10 scale with categories

| Score Range | Risk Rating | Recommendation |
|-------------|-------------|----------------|
| 0-2 | LOW | APPROVE |
| 3-4 | MODERATE | APPROVE WITH CONDITIONS |
| 5-7 | HIGH | REJECT |
| 8-10 | CRITICAL | REJECT |

Components:
| Component | Points | Criteria |
|-----------|--------|----------|
| Performance | 3 max | AUC < 0.55 = +3, AUC 0.55-0.60 = +2, AUC 0.60-0.65 = +1 |
| Stability | 2 max | PSI > 0.25 = +2, PSI 0.15-0.25 = +1 |
| Calibration | 2 max | HL p-value ≤ 0.05 = +2 |
| Fairness | 3 max | DI < 0.6 = +3, DI 0.6-0.8 = +1 |
| Leakage | 3 max | Leakage detected = +3 |
"""

# Calculate risk score
risk_score = 0
risk_factors = []

# Performance (V2.0: updated for realistic AUC)
if best_auc < 0.55:
    risk_score += 3
    risk_factors.append(f"AUC-ROC {best_auc:.3f} below 0.55")
elif best_auc < 0.6:
    risk_score += 2
    risk_factors.append(f"AUC-ROC {best_auc:.3f} between 0.55 and 0.6")
elif best_auc < 0.65:
    risk_score += 1
    risk_factors.append(f"AUC-ROC {best_auc:.3f} between 0.6 and 0.65")
else:
    risk_factors.append(f"AUC-ROC {best_auc:.3f} above 0.65")

# Stability
if psi is not None:
    if psi > 0.25:
        risk_score += 2
        risk_factors.append("PSI > 0.25 (major shift)")
    elif psi > 0.15:
        risk_score += 1
        risk_factors.append("PSI between 0.15 and 0.25 (moderate shift)")
    else:
        risk_factors.append("PSI stable (< 0.15)")

# Calibration
poor_calibration = [r for r in calibration_results if r['HL_p_value'] <= 0.05]
if poor_calibration:
    risk_score += 2
    risk_factors.append(f"Poor calibration: {len(poor_calibration)} models")
else:
    risk_factors.append("Calibration acceptable")

# Fairness (V2.0 - CRITICAL)
if not fairness_df.empty:
    min_di = fairness_df['DI_Ratio'].min()
    if min_di < 0.6:
        risk_score += 3
        risk_factors.append(f"SEVERE disparate impact (DI: {min_di:.4f}) - DEPLOYMENT BLOCKED")
    elif min_di < 0.8:
        risk_score += 1
        risk_factors.append(f"Disparate impact concerns (DI: {min_di:.4f})")
    else:
        risk_factors.append("No disparate impact detected")

# Leakage (V2.0: addressed)
leakage_found = False
if leakage_found:
    risk_score += 3
    risk_factors.append("Data leakage detected")
else:
    risk_factors.append("No data leakage detected")

# Determine rating
if risk_score >= 8:
    risk_rating = "CRITICAL"
    status = "REJECT"
elif risk_score >= 6:
    risk_rating = "HIGH"
    status = "REJECT"
elif risk_score >= 4:
    risk_rating = "MODERATE-HIGH"
    status = "CONDITIONAL"
elif risk_score >= 2:
    risk_rating = "MODERATE"
    status = "APPROVE WITH MONITORING"
else:
    risk_rating = "LOW"
    status = "APPROVE"

print(f"\nRisk Score: {risk_score}/10")
print(f"Risk Rating: {risk_rating}")
print(f"Recommendation: {status}")
```
